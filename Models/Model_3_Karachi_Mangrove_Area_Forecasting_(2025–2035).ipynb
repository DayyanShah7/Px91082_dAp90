Model 3: Karachi Mangrove Area Forecasting (2025–2035)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

X = master_df[["year"]]
y = master_df["mangrove_area_ha"]

mangrove_model = LinearRegression()
mangrove_model.fit(X, y)

y_pred = mangrove_model.predict(X)

r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("R² Score:", round(r2,4))
print("MAE:", round(mae,4))
print("RMSE:", round(rmse,4))

future_years = pd.DataFrame({
    "year": np.arange(2025,2036)
})

future_mangrove = mangrove_model.predict(future_years)

forecast_mangrove = pd.DataFrame({
    "year": future_years["year"],
    "predicted_mangrove_area_ha": future_mangrove
})

plt.figure(figsize=(12,6))

plt.scatter(
    master_df["year"],
    master_df["mangrove_area_ha"],
    label="Historical Data"
)

plt.plot(
    master_df["year"],
    y_pred,
    label="Linear Regression Trend"
)

plt.plot(
    future_years["year"],
    future_mangrove,
    "--",
    label="Forecast (2025-2035)"
)

plt.title("Karachi Mangrove Area Forecast (2000-2035)")
plt.xlabel("Year")
plt.ylabel("Mangrove Area (ha)")
plt.legend()
plt.grid(True)

plt.show()

forecast_mangrove.head()
```
