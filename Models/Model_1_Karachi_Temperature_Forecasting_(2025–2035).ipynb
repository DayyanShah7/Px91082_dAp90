Model 1: Karachi Temperature Forecasting (2025–2035)

```python
import pandas as pd

climate = pd.read_excel("Karachi_Climate.xlsx")
forest = pd.read_excel("Karachi_Forest.xlsx")
mangrove = pd.read_csv("Karachi_Mangrove.csv")
master_df = pd.read_csv("karachi_master_dataset.csv")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# Features and target
X = master_df[["year"]]
y = master_df["avg_temperature"]

# Train model
temp_model = LinearRegression()
temp_model.fit(X, y)

# Predictions on historical data
y_pred = temp_model.predict(X)

# Evaluation
r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("R² Score:", round(r2, 4))
print("MAE:", round(mae, 4))
print("RMSE:", round(rmse, 4))
```

```python
future_years = pd.DataFrame({
    "year": np.arange(2025, 2036)
})

future_temp = temp_model.predict(future_years)

forecast_temp = pd.DataFrame({
    "year": future_years["year"],
    "predicted_temperature": future_temp
})

forecast_temp
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

# Historical data
plt.scatter(
    master_df["year"],
    master_df["avg_temperature"],
    label="Historical Data"
)

# Trend line over historical period
plt.plot(
    master_df["year"],
    y_pred,
    label="Linear Regression Trend"
)

# Forecast
plt.plot(
    future_years["year"],
    future_temp,
    linestyle="--",
    label="Forecast (2025-2035)"
)

plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.title("Karachi Temperature Forecast (2000-2035)")
plt.legend()
plt.grid(True)

plt.show()
```

```python
forecast_temp.to_csv(
    "temperature_forecast_2025_2035.csv",
    index=False
)
```
