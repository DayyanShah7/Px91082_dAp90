```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

X = master_df[["year"]]
y = master_df["tree_cover_loss_ha"]

forest_model = LinearRegression()
forest_model.fit(X, y)

y_pred = forest_model.predict(X)

r2 = r2_score(y, y_pred)
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))

print("R² Score:", round(r2,4))
print("MAE:", round(mae,4))
print("RMSE:", round(rmse,4))

future_years = pd.DataFrame({
    "year": np.arange(2025,2036)
})

future_forest_loss = forest_model.predict(future_years)

forecast_forest = pd.DataFrame({
    "year": future_years["year"],
    "predicted_tree_cover_loss_ha": future_forest_loss
})

plt.figure(figsize=(12,6))

plt.scatter(master_df["year"],
            master_df["tree_cover_loss_ha"],
            label="Historical Data")

plt.plot(master_df["year"],
         y_pred,
         label="Linear Regression Trend")

plt.plot(future_years["year"],
         future_forest_loss,
         "--",
         label="Forecast (2025-2035)")

plt.title("Karachi Tree Cover Loss Forecast (2000-2035)")
plt.xlabel("Year")
plt.ylabel("Tree Cover Loss (ha)")
plt.legend()
plt.grid(True)
plt.show()

forecast_forest.head()
```
