Exploratory Data Analysis (EDA)

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

Climate EDA

```python
yearly_temp = climate.groupby("year")["tavg"].mean()

plt.figure(figsize=(12,6))
plt.plot(yearly_temp.index, yearly_temp.values)
plt.title("Karachi Average Temperature Trend")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
yearly_rain = climate.groupby("year")["prcp"].sum()

plt.figure(figsize=(12,6))
plt.plot(yearly_rain.index, yearly_rain.values)
plt.title("Karachi Rainfall Trend")
plt.show()
```
