Environmental Risk Index

```python
import pandas as pd

master_df = pd.read_csv(
    "karachi_master_dataset.csv"
)

master_df.head()
```

```python
Index(['year', 'avg_temperature', 'annual_rainfall', 'tree_cover_loss_ha',
       'tree_cover_gain_ha', 'mangrove_area_ha'],
      dtype='object')
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = master_df[["year"]]
y = master_df["avg_temperature"]

temp_model = LinearRegression()
temp_model.fit(X, y)

future_years = pd.DataFrame({
    "year": np.arange(2025,2036)
})

future_temp = temp_model.predict(
    future_years
)

forecast_temp = pd.DataFrame({
    "year": future_years["year"],
    "predicted_temperature": future_temp
})
```

```python
X = master_df[["year"]]
y = master_df["tree_cover_loss_ha"]

forest_model = LinearRegression()
forest_model.fit(X, y)

future_forest = forest_model.predict(
    future_years
)

forecast_forest = pd.DataFrame({
    "year": future_years["year"],
    "predicted_tree_cover_loss_ha": future_forest
})
```

```python
X = master_df[["year"]]
y = master_df["mangrove_area_ha"]

mangrove_model = LinearRegression()
mangrove_model.fit(X, y)

future_mangrove = mangrove_model.predict(
    future_years
)

forecast_mangrove = pd.DataFrame({
    "year": future_years["year"],
    "predicted_mangrove_area_ha": future_mangrove
})
```

```python
future_df = forecast_temp.merge(
    forecast_forest,
    on="year"
)

future_df = future_df.merge(
    forecast_mangrove,
    on="year"
)

future_df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

future_df[
    [
        "temp_score",
        "forest_score",
        "mangrove_score"
    ]
] = scaler.fit_transform(
    future_df[
        [
            "predicted_temperature",
            "predicted_tree_cover_loss_ha",
            "predicted_mangrove_area_ha"
        ]
    ]
) * 100
```

```python
future_df["risk_score"] = (
    0.40 * future_df["temp_score"]
    +
    0.35 * future_df["forest_score"]
    +
    0.25 * (100 - future_df["mangrove_score"])
)
```

```python
def classify_risk(score):

    if score < 25:
        return "Low"

    elif score < 50:
        return "Moderate"

    elif score < 75:
        return "High"

    else:
        return "Critical"

future_df["risk_category"] = (
    future_df["risk_score"]
    .apply(classify_risk)
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(
    future_df["year"],
    future_df["risk_score"],
    marker="o"
)

plt.title(
    "Karachi Environmental Risk Forecast (2025-2035)"
)

plt.xlabel("Year")
plt.ylabel("Risk Score")

plt.grid(True)

plt.show()
```

```python
future_df.to_csv(
    "karachi_environmental_risk_forecast.csv",
    index=False
)
```
